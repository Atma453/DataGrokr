```python
import pandas as pd
from datetime import datetime

class BankAccount:
    def __init__(self, account_number, holder_name, balance=0):
        self.account_number = account_number
        self.holder_name = holder_name
        self.balance = balance
        self.transactions = []

    def deposit(self, amount):
        if amount <= 0:
            print("Invalid deposit amount")
            return
        self.balance += amount
        self.transactions.append({
            "Date": datetime.now().strftime("%Y-%m-%d %H:%M:%S"),
            "Type": "Deposit",
            "Amount": amount,
            "Balance": self.balance
        })
        print(f"₹{amount:.2f} deposited successfully")

    def withdraw(self, amount):
        if amount <= 0:
            print("Invalid withdrawal amount")
            return
        if amount > self.balance:
            print("Insufficient balance")
            return
        self.balance -= amount
        self.transactions.append({
            "Date": datetime.now().strftime("%Y-%m-%d %H:%M:%S"),
            "Type": "Withdrawal",
            "Amount": amount,
            "Balance": self.balance
        })
        print(f"₹{amount:.2f} withdrawn successfully")

    def check_balance(self):
        print(f"Current Balance: ₹{self.balance:.2f}")

    def show_transactions(self):
        if not self.transactions:
            print("No transactions available")
            return

        df = pd.DataFrame(self.transactions)
        print("\nTransaction History")
        print(df.to_string(index=False))

    def save_transactions(self, filename="transactions.csv"):
        if self.transactions:
            df = pd.DataFrame(self.transactions)
            df.to_csv(filename, index=False)
            print(f"Transactions saved to {filename}")
        else:
            print("No transactions to save")


def analyze_csv(filename="transactions.csv"):
    try:
        df = pd.read_csv(filename)

        print("\nCSV Dataset")
        print(df.to_string(index=False))

        print("\nDataset Information")
        print(df.info())

        print("\nTotal Transactions:", len(df))
        print("Total Deposits:", df[df["Type"] == "Deposit"]["Amount"].sum())
        print("Total Withdrawals:", df[df["Type"] == "Withdrawal"]["Amount"].sum())
        print("Average Transaction:", df["Amount"].mean())
        print("Maximum Transaction:", df["Amount"].max())
        print("Minimum Transaction:", df["Amount"].min())

        print("\nTransaction Summary")
        print(df.groupby("Type")["Amount"].agg(["count", "sum", "mean"]))

    except FileNotFoundError:
        print("CSV file not found")


def main():
    account = BankAccount("ACC1001", "Atmanand", 5000)

    while True:
        print("\n===== OOP BANK ACCOUNT =====")
        print("1. Deposit")
        print("2. Withdraw")
        print("3. Check Balance")
        print("4. Transaction History")
        print("5. Save Transactions")
        print("6. Analyze CSV")
        print("7. Exit")

        choice = input("Enter your choice: ")

        if choice == "1":
            amount = float(input("Enter deposit amount: "))
            account.deposit(amount)

        elif choice == "2":
            amount = float(input("Enter withdrawal amount: "))
            account.withdraw(amount)

        elif choice == "3":
            account.check_balance()

        elif choice == "4":
            account.show_transactions()

        elif choice == "5":
            account.save_transactions()

        elif choice == "6":
            filename = input("Enter CSV filename: ")
            analyze_csv(filename)

        elif choice == "7":
            print("Thank you for using the banking system")
            break

        else:
            print("Invalid choice")


if __name__ == "__main__":
    main()
```
